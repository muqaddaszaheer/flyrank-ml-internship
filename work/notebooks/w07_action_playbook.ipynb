## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

Before taking action, a human should review the page context, the available performance evidence, the reason code, and whether the recommendation makes sense for the content.

Human review is required before changing, removing, redirecting, or rewriting content.

### No-go list

The system should never automatically:

- publish content;
- delete content;
- redirect URLs;
- make major content changes;
- make decisions about sensitive or high-impact content;
- treat a model prediction as proof of future performance.

The model should remain a prioritization and decision-support tool.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
no_go_actions = [
    "automatic publishing",
    "automatic deletion",
    "automatic URL redirects",
    "automatic major content changes",
    "automatic high-impact decisions"
]

print("Human review required: YES")
print("No-go actions:", len(no_go_actions))

for action in no_go_actions:
    print("-", action)

print("\nPASS: recommendations require human review.")
```

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

The recommendations should be monitored because model performance and content conditions can change over time.

Useful warning signs include:

- measured model performance decreases;
- the distribution of important features changes;
- recommendation patterns change substantially;
- the available content data changes;
- error examples become more common;
- reason codes no longer match the observed data.

A retraining or review should be considered when the measured performance or data distribution shows a meaningful change.

Monitoring should be treated as a review trigger rather than an automatic retraining decision.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
monitoring_triggers = [
    "model performance decreases",
    "feature distribution changes",
    "recommendation patterns change",
    "data conditions change",
    "error rate increases",
    "reason codes become less useful"
]

print("Monitoring triggers:")

for trigger in monitoring_triggers:
    print("-", trigger)

print("\nPASS: monitoring and retrain triggers defined.")
```

## 5. Exports for the paper

I export a simple ranked action queue for the paper. The queue contains the recommended action, reason code, priority, and a short explanation.

The exported file is intended for research and paper use. It is not a production system and does not automatically change or publish content.

```python
from pathlib import Path
import pandas as pd

# Set repository and output folder
repo_dir = Path("/content/flyrank-ml-internship")
output_dir = repo_dir / "work" / "outputs"
output_dir.mkdir(parents=True, exist_ok=True)

# Create ranked action queue
action_queue = pd.DataFrame([
    {
        "priority": 1,
        "action": "Refresh declining content",
        "reason_code": "declining_performance",
        "explanation": "Prioritize pages showing evidence of declining performance."
    },
    {
        "priority": 2,
        "action": "Review weak engagement",
        "reason_code": "low_engagement",
        "explanation": "Review pages with weaker observed engagement signals."
    },
    {
        "priority": 3,
        "action": "Review weak click-through performance",
        "reason_code": "low_ctr",
        "explanation": "Review pages with weaker observed click-through performance."
    },
    {
        "priority": 4,
        "action": "Review older content",
        "reason_code": "content_age",
        "explanation": "Review older pages as part of a broader content review."
    },
    {
        "priority": 5,
        "action": "Monitor pages without a clear action signal",
        "reason_code": "insufficient_evidence",
        "explanation": "Do not force an action when the available evidence is weak."
    }
])

# Export for the paper
queue_path = output_dir / "action_queue.csv"
action_queue.to_csv(queue_path, index=False)

print("Action queue exported successfully.")
print("File:", queue_path)
print("File exists:", queue_path.exists())

display(action_queue)
```
